```python
# Inbuilt Function to check if Input is Alphanumeric
a1 = input("Enter a String: ")
if a1.isalnum() == True:
    print("Valid")
else:
    print("Invalid")
```

```python
# Imported RegEx module to check if Input is Alphanumeric (findall)
import re
txt = input("Enter a String to check its validity as and alphanumeric string : ")
x = re.findall(r'\^[a-zA-Z0-9]+$', txt)
if x :
    print("Invalid")
else:
    print("Valid")
```

```python
#Imported RegEx module to check if Input is Alphanumeric (fullmatch)
import re
def is_alphanumeric(S):
    return bool(re.fullmatch(r'[A-Za-z0-9]+',S))
txt = input("Enter a String")
if is_alphanumeric(txt):
    print("Valid")
else:
    print("Invalid")
```
