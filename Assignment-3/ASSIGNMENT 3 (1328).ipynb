```python
# Python Code to Check if Triangle is a Right Angled Triangle.
def t(s1, s2,s3):
    if (s1)**2 + (s2)**2 == (s3)**2 or (s1)**2 + (s3)**2 == (s2)**2 or (s2)**2 + (s3)**2 == (s1)**2 :
        print ("The Triangle is a Right Angled Triangle.")
    else :
        print ("The Triangle is Not a Right Angled Triangle.")

s1 = int(input("Enter length of 1st side : "))
s2 = int(input("Enter length of 2nd side : "))
s3 = int(input("Enter length of 3rd side : "))

t(s1, s2, s3)
```
